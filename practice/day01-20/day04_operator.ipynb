```python
s = input('输入华氏度')
f = float(s)
c = (f - 32) / 1.8
print('%.1f华氏度 = %.1f摄氏度' %(f, c))
print(f'{f:.1f}华氏度 = {c:.1f}摄氏度')
```

```python
"""
输入半径，计算圆的周长和面积
"""

import math

s = input('输入圆的半径')
r =float(s)

perimeter = 2 * math.pi * r
area = math.pi * (r ** 2)

print(f'圆的周长是{perimeter:.2f}')
print(f'圆的面积是{area:.2f}')
```

```python
""""
输入年份，判断闰年

"""

year = int(input('输入年份：'))

is_leap = year % 4 == 0 and year % 100 != 0 or year % 400 == 0

print(f'{year}是闰年吗', {is_leap})
```
