```python
"""
BMI Calculator

BMI normal range:   18.5 <= BMI <= 24
fat:    BMI >= 27
"""
h = float(input('输入你的身高/m:'))
w = float(input('输入你的体重/kg:'))

BMI = w / (h ** 2)

print(f'你的BMI为{BMI:.2f}')

if 18.5 <= BMI <= 24:
    print('你的身材中中中！！！')
else:
    print('你的身材不太中')

```

```python
"""
BMI Calculator 进阶版
"""

h = float(input('输入你的身高/m:'))
w = float(input('输入你的体重/kg'))

BMI = w / h ** 2

print(f'你的BMI为{BMI:.2f}')

if BMI < 18.5:
    print('你的身材过轻')
elif BMI < 24:
    print('你的身材正好')
elif BMI < 27:
    print('你的身材过重')
elif BMI < 30:
    print('你已轻度肥胖')
elif BMI < 35:
    print('你已中度肥胖')
else:
    print('你已重度肥胖')
```

```python
status_code = int(input('status_code:'))

match status_code:
    case 400: description = 'Bad Quest'
    case 401: description = 'unauthorized'
    case 403: description = 'forbidden'
    case 404: description = 'not found'
    case 405: description = 'method not allowed'
    case 418: description = 'i am teapot'
    case 429: description = 'too many requests'
    case _: description = 'unknown status code'
print(f'状态码描述: {description}')
```

```python
"""
分段函数求值
"""

x = float(input('输入自变量x的值:'))

if x > 1:
    y = 3 * x - 5
elif -1 <= x <= 1:
    y = x + 2
elif x < -1:
    y = 5 * x + 3
print(f'y ={y:.2f}')
```

```python
score = float(input('输入考生成绩:'))

if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
elif score >= 60:
    grade = 'D'
else:
    grade = 'E'
print(f'考生等级为{grade}')
```

```python
"""
计算三角形的周长和面积
"""

a = float(input('输入第一条边长'))
b = float(input('输入第二条边长'))
c = float(input('输入第三条边长'))

if a + b > c and a + c > b and b + c > a:
    perimeter = a + b + c
    
    s = perimeter / 2
    area = (s * (s - a) * (s - b) * (s - c)) ** 0.5
    
    print(f'该三角形周长为{perimeter:.2f}, 面积为{area:.2f}')
else:
    print('这组边长不能构成三角形')
```
