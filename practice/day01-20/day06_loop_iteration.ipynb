```python
"""
1-100 整数求和
"""
total = 0
for i in range(1, 101):
    total += i
print(f'1到100的整数求和值为{total}')
    
```

```python
"""
1到100 偶数求和 for
"""

total = 0
for i in range(1, 101):
    if i % 2 == 0:
        total += i
print(f'1-100的偶数求和值为{total}')

total = 0
for i in range(2, 101, 2):
    total += i
print(f'1-100的偶数求和值为{total}')
```

```python
"""
1到100 整数求和 while
"""
total = 0
i = 1
while i <= 100:
    total += i
    i += 1
print(f'1到100整数while求和为{total}')
```

```python
"""
1到100 偶数求和 while
"""

total = 0
i = 2
while i <= 100:
    total += i
    i += 2
print(f'1-100偶数求和值为{total}')
```

```python
i = 2
total = 0
while True:
    total += i
    i += 2
    if i > 100:
        break
print(total)
```

```python
total = 0

for i in range(1, 101):
    if i % 2 != 0:
        continue
    total += i
print(total)
```

```python
"""
乘法口诀表
"""
for i in range(1, 10):
    for j in range(1, i + 1):
        print(f'{i} * {j} = {i * j}', end = '\t')
    print()
```

```python
"""
判断素数 Prime number
"""
num = int(input('输入一个数来判断是否为素数Prime number:'))
end = int(num ** 0.5)
is_prime = True

for i in range(2, end + 1):
    if num % i == 0:
        is_prime = False
        break
    
if is_prime:
    print(f'{num}是质数')
else:
    print(f'{num}不是质数')
```

```python
"""
两个数的最大公约数
"""
a = int(input('输入第一个数'))
b = int(input('输入第二个数'))

for i in range(a, 0, -1):
    if a % i == 0 and b % i == 0:
        print(f'最大公约数{i}')
        break
```

```python
import random

answer = random.randrange(1, 101)
counter = 0

while True:
    num = int(input('输入: '))
    if num > answer:
        print(f'{num}大了')
    elif num < answer:
        print(f'{num}小了')
    else:
        print(f'{num}中了')
        break
    counter += 1
print(f'你一共猜了{counter}次')
```
