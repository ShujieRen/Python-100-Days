```python
"""
随机验证码：设计一个生成随机验证码的函数，验证码由数字和英文大小写字母构成，长度可以通过参数设置。
"""
import random
import string

ALL_CHARS = string.digits + string.ascii_letters

def generate_code(*, code_length=4):
    """
    生成制定长度的随机验证码
    :param code_length: 验证码的长度，默认4个字符
    :return: 由大小写英文字母和数字组成的随机验证码字符串
    """
    return ''.join(random.choices(ALL_CHARS, k=code_length))
```

```python
for _ in range(5):
    print(generate_code())

for _ in range(5):
    print(generate_code(code_length=6))

for _ in range(5):
    print(generate_code(code_length=8))
```

```python
"""
判断素数
"""
def is_prime(num: int) -> bool:
    """
    判断一个正整数是否为素数
    :param num: 大于1的正整数
    :return: 如果num是质数就返回True, 否则返回False
    """
    if num < 2:
        return False
    for i in range(2, int(num ** 0.5)+1):
        if num % i == 0:
            return False
    return True
```

```python
for i in range(10):
    print(f'{i}, {is_prime(i)}')
```

```python
"""
最大公约数和最小公倍数
"""
def lcm(x: int, y: int) -> int:
    """
    求最小公倍数
    :param x, y: 两个整数
    :return: 最小公倍数
    """
    return x * y // gcd(x, y)


def gcm(x: int, y: int) -> int:
    """
    求最大公约数
    :param: x, y: 两个整数
    :return: 最大公约数
    """
    while y % x != 0:
        x, y = y % x, x
    return x
```

```python
"""
数据统计
"""
def ptp(data):
    """
    极差
    """
    return max(data) - min(data)

def mean(data):
    """
    算术平均
    """
    return sum(data) / len(data)

def median(data):
    """
    中位数
    """
    length = len(data)
    temp = sorted(data)

    if length % 2 == 0:
        m = (temp[length // 2] + temp[length // 2 - 1]) / 2
    else:
        m = temp[length // 2]
    return m


def var(data, ddof=1):
    """方差"""
    x_bar = mean(data)
    temp = [(num - x_bar) ** 2 for num in data]
    return sum(temp) / (len(temp) - ddof)


def std(data, ddof=1):
    """标准差"""
    return var(data, ddof) ** 0.5

def cv(data, ddof=1):
    """变异系数"""
    return std(data, ddof) / mean(data)


def describe(data):
    """输出描述性统计信息"""
    print(f'均值: {mean(data)}')
    print(f'中位数: {median(data)}')
    print(f'极差: {ptp(data)}')
    print(f'方差: {var(data)}')
    print(f'标准差: {std(data)}')
    print(f'变异系数: {cv(data)}')
```

```python
"""
双色球随机选号程序
"""
import random

red_balls = [i for i in range(1, 34)]
blue_balls = [i for i in range(1, 17)]

def choose():
    """
    生成一组随机号码
    :return: 保存随机号码的列表
    """
    selected_balls = random.sample(red_balls, 6)
    selected_balls.sort()
    selected_balls.append(random.choice(blue_balls))
    return selected_balls

def display(balls):
    """
    格式输出一组号码
    :param balls: 保存的随机号码列表
    :return: 格式化的保存的随机号码列表
    """
    for ball in balls[:-1]:
        print(f'\033[031m{ball:0>2d}\033[0m', end=' ')
    print(f'\033[034m{balls[-1]:0>2d}\033[0m')

n = int(input('打几注号码:'))
for _ in range(n):    
    display(choose())
```
