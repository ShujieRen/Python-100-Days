```python
"""
统计6000次骰子的分布
"""

import random

f1 = 0
f2 = 0
f3 = 0
f4 = 0
f5 = 0
f6 = 0

for _ in range(6000):
    dice = random.randrange(1, 7)
    if dice == 1:
        f1 += 1
    elif dice == 2:
        f2 += 1       
    elif dice == 3:
        f3 += 1        
    elif dice == 4:
        f4 += 1  
    elif dice == 5:
        f5 += 1
    elif dice == 6:
        f6 += 1

print(f'有{f1}次1')
print(f'有{f2}次2')
print(f'有{f3}次3')
print(f'有{f4}次4')
print(f'有{f5}次5')
print(f'有{f6}次6')
```

```python
frequency = [0, 0, 0, 0, 0, 0]

for _ in range(6000):
    dice = random.randrange(1, 7),
    if dice == 1:
        frequency[0] += 1
    if dice == 2:
        frequency[1] += 1
    if dice == 3:
        frequency[2] += 1
    if dice == 4:
        frequency[3] += 1
    if dice == 5:
        frequency[4] += 1
    if dice == 6:
        frequency[5] += 1
```

```python
import random

frequency = [0] * 6

for _ in range(6000):
    dice = random.randrange(1, 7)
    frequency[dice - 1] += 1
    
for i in range(1, 7):
    print(f'{i}出现的次数为{frequency[i - 1]}')
```
