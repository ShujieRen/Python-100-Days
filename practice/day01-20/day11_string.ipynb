```python
s1 = 'hello world'
s2 = '''
hello
world
!!!
'''
print(s1)
print(s2)
```

```python
s1 = '\'hello world\''
s2 = '\\hello, world!\\'

print(s1)
print(s2)
```

```python
s1 = '\it \is \time \to \read \now'
s2 = r'\it \is \time \to \read \now'
print(s1)
print(s2)
```

```python
s1 = '\141\142\143\x61\x62\x63'
s2 = '\u9a86\u660a'
print(s1)
print(s2)
```

```python
s1 = 'hello' + ', ' + 'world'

print(s1)

s2 = '!'
print(s2)

s3 = s1 + s2 * 3
print(s3)
```

```python
s1 = 'hello world'
print(s1.capitalize())
print(s1.title())
print(s1.upper())

s2 = 'GOODBYE'
print(s2.lower())
```
