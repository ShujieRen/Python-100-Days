```python
"""
阶乘
"""
def factorial(num):
    result = 1
    for i in range(2, num + 1):
        result *= i
    return result

m = int(input('m = '))
n = int(input('n ='))

print(factorial(m) // factorial(n) // factorial(m - n))
```
